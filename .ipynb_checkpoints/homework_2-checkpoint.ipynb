```python
# task 1
def sum_distance(num_1: int, num_2: int) -> int:
    return (num_1 + num_2) * (abs(num_2 - num_1) + 1) // 2

print(sum_distance(1, 100))
print(sum_distance(100, 1))
```

```python
# task 2
def trim_and_repeat(text: str, offset: int = 0, repetitions: int = 1) -> str:
    return text[offset:] * repetitions

print(trim_and_repeat('123456', 3, 3))
print(trim_and_repeat('123456', 3))
print(trim_and_repeat('123456', repetitions=3))
print(trim_and_repeat('123456'))
```
