```python
# task 1
word: str = '12345'
word_len: int = len(word)
word_len_mid: int =  word_len // 2

print(word[word_len_mid] if word_len % 2 != 0 else word[word_len_mid: word_len_mid + 1])

word: str = '123456'
word_len: int = len(word)
word_len_mid: int =  word_len // 2

print(word[word_len_mid] if word_len % 2 != 0 else word[word_len_mid - 1: word_len_mid + 1])
```

```python
# task 2
boys: list[str] = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls: list[str] = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

boys.sort()
girls.sort()

warning_text: str = 'Внимание, кто-то может остаться без пары!'

if len(boys) == len(girls):
    print('\n'.join([f'{boy} и {girl}' for boy, girl in zip(boys, girls)]))
else:
    print(warning_text)

boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Michael']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) == len(girls):
    print('\n'.join([f'{boy} и {girl}' for boy, girl in zip(boys, girls)]))
else:
    print(warning_text)
```
