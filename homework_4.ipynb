```python
# task 1
from datetime import datetime

the_moscow_times: str = 'Wednesday, October 2, 2002'
the_guardian: str = 'Friday, 11.10.13'
daily_news: str = 'Thursday, 18 August 1977'

dt_moscow_times = datetime.strptime(the_moscow_times, "%A, %B %d, %Y")
dt_guardian = datetime.strptime(the_guardian, "%A, %d.%m.%y")
dt_daily_news = datetime.strptime(daily_news, "%A, %d %B %Y")

print(dt_moscow_times, dt_guardian, dt_daily_news, sep='\n')
```

```python
# task 2
from datetime import datetime, timedelta

def date_range(start_date: str, end_date: str) -> list[str]:
    try:
        start_dt: datetime = datetime.strptime(start_date, '%Y-%m-%d')
        end_dt: datetime = datetime.strptime(end_date, '%Y-%m-%d')

    except ValueError:
        return []

    dates: list[str] = []
    while start_dt <= end_dt:
        dates.append(start_dt.strftime('%Y-%m-%d'))
        start_dt += timedelta(days=1)

    return dates

res: list[str] = date_range('2022-01-01', '2022-01-03')
print(res)
res = date_range('2022-02-30', '2022-02-31')
print(res)

```
