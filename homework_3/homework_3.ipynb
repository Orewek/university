```python
# task 1
import json

purchases: dict[str, str] = {}
with open('purchase_log.txt', 'r', encoding='utf-8') as log_f:
    for line in log_f:
        line_data: dict[str, str] = json.loads(line.strip())

        if line_data['user_id'] == 'user_id':
            continue

        purchases[line_data['user_id']] = line_data['category']

for user_id, category in list(purchases.items())[:2]:
    print(f"{user_id} '{category}'")
```

```python
# task 2
# note: В .zip лежал visit_log.xlsx, конвертнул в .csv
import csv

with open('funnel.csv', 'w', encoding='utf-8') as funnel_f:
    writer = csv.writer(funnel_f)
    writer.writerow(['user_id', 'source', 'category'])
    
    with open('visit_log.csv', 'r', encoding='utf-8') as visit_f:
        for line in visit_f:
            row: list[str] = line.strip().split(',')

            user_id: str = row[0].strip('"')
            source: str = row[1].strip('"')
            
            if row[0] == 'user_id':
                continue

            if user_id in purchases:
                category: str = purchases[user_id]
                writer.writerow([user_id, source, category])
                

with open('funnel.csv', 'r', encoding='utf-8') as f:
    for i in range(5):
        print(f.readline().strip())
```
